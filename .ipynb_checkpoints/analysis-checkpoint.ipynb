```python
import pandas as pd
from scipy.stats import linregress
import matplotlib.pyplot as plt
```

```python

raw = pd.read_csv("data.csv", encoding="cp1252", header=None)
raw.head()

```

```python
with pd.option_context("display.max_rows", None, "display.max_colwidth", None):
    display(raw[0])
```

```python
raw.shape
raw.describe()
```

```python

starts = raw.index[
    raw[0].fillna("").str.startswith("OFCOM MEDIA LITERACY")
].tolist()

sections = []

for i, start in enumerate(starts):
    end = starts[i + 1] if i + 1 < len(starts) else len(raw)

    sections.append({
        "start": start,
        "end": end,
        "question": raw.iloc[start + 3, 0]
    })

sections = pd.DataFrame(sections)
print(sections.to_string())
```

```python
import seaborn as sns
```

```python
usage = raw.iloc[[863, 866, 869, 872, 875], 2:5].copy()
usage.index = ["TV", "Tablet", "Phone", "Computer", "Games console"]
usage.columns = ["Age 5–7", "Age 8–11", "Age 12–15"]

for column in usage.columns:
    usage[column] = usage[column].str.rstrip("%").astype(float)
    
plt.figure(figsize=(7, 4))
sns.heatmap(
    usage, annot=True, fmt=".0f",
    cmap="Reds", vmin=0, vmax=100,
    cbar_kws={"label": "devices kid use (%)"}
        )
plt.title("devices used to watch programs and films")
plt.tight_layout()
plt.show()
```

```python
# kids uses more than 1 devices, which the percentages added up > 100%
```

```python

normalized = usage.div(usage.sum(axis=0), axis=1) * 100
sns.heatmap(
    normalized, annot=True, fmt=".1f",
    cmap="Reds", vmin=0, vmax=100,
    cbar_kws={"label": "share of devices (%)"}
    )
plt.show()
```

```python

from scipy.stats import linregress

ages = [6, 9.5, 13.5]
phone_use = [11, 14, 33]

result = linregress(ages, phone_use)

plt.scatter(ages, phone_use)
plt.plot(
    ages,
    [result.intercept + result.slope * age for age in ages],
    color="red"
)
plt.xlabel("Age-group midpoint")
plt.ylabel("Watching TV or films on a phone (%)")
plt.title("Phone viewing by age group — QP5")
plt.tight_layout()
plt.show()

print(f"Slope: {result.slope:.2f} percentage points per year")
print(f"R²: {result.rvalue**2:.3f}")
```
