```python
age_label = ["Age 5–7", "Age 8–11", "Age 12–15"]
def get_age_percentages(code):
    
    table = get_section(code)

    labels = table[0].replace(r"^\s*$", pd.NA, regex=True).ffill()

    is_percentage = (
        table[1].astype("string").str.endswith("%", na=False)
    )

    result = table.loc[is_percentage, [2, 3, 4]].copy()
    result.index = labels.loc[is_percentage].tolist()
    result.index.name = "Response"
    result.columns = age_label

    for column in result.columns:
        result[column] = pd.to_numeric(
            result[column].astype("string").str.rstrip("%"),
            errors="coerce"
        )

    return result


devices = get_age_percentages("QP5")
attitudes = get_age_percentages("QC4A1")

display(devices)
display(attitudes)
```

```python
def get_all_percentages(code):
    table = get_section(code)

    # Locate the header row containing "Total".
    header_mask = table[1].astype("string").str.strip().eq("Total")
    header_position = next(
        (i for i, match in enumerate(header_mask.fillna(False)) if match),
        None
    )

    if header_position is None:
        raise ValueError(f"No subgroup headers found for {code}")

    # Keep original column numbers so repeated labels remain distinguishable.
    headers = table.iloc[header_position, 1:]
    columns = pd.MultiIndex.from_tuples(
        [
            (column, str(label).strip())
            for column, label in headers.items()
        ],
        names=["Column", "Subgroup"]
    )

    labels = table[0].replace(
        r"^\s*$", pd.NA, regex=True
    ).ffill()

    is_percentage = (
        table[1].astype("string").str.endswith("%", na=False)
    )

    result = table.loc[is_percentage].iloc[:, 1:].copy()
    result.index = labels.loc[is_percentage].tolist()
    result.index.name = "Response"
    result.columns = columns

    for column in result.columns:
        result[column] = pd.to_numeric(
            result[column].astype("string").str.rstrip("%"),
            errors="coerce"
        )

    return result
```

- H₀₁: Perceived regional representation is independent of household social class.

```python
def list_subgroups(code):
    table = all_question_data[code]
    return table.columns.to_frame(index=False)

display(list_subgroups("QC4C1"))
```

```python
def select_subgroups(code, column_numbers):
    table = all_question_data[code]

    mask = table.columns.get_level_values("Column").isin(column_numbers)

    if not mask.any():
        raise ValueError("None of these column numbers were found")

    result = table.loc[:, mask].copy()
    result.columns = result.columns.get_level_values("Subgroup")

    return result
```

```python
social_class_columns = [14, 15, 16, 17]  
location_columns = [21, 22] 
```

```python
def compare_representation(code, column_numbers):
    table = select_subgroups(code, column_numbers)

    result = table.loc[["Yes", "No", "Don't know"]].T
    result.index.name = "Subgroup"

    return result.sort_values("No", ascending=False)
```

```python
def compare_importance(code, column_numbers):
    table = select_subgroups(code, column_numbers)

    result = table.loc[
        ["Important", "Neither", "Not important", "Don't know"]
    ].T
    result.index.name = "Subgroup"

    return result.sort_values("Important", ascending=False)



def find_sections(code):
    return sections.loc[
        sections["question"].str.startswith(f"{code}.", na=False),
        ["start", "end", "question"]
    ].copy()
```

```python
representation_questions = ["QC4B1", "QC4C1", "QC4D1"]

subgroup_columns = {
    "Social grade": [14, 15, 16, 17],
    "Location": [21, 22]
}

representation_results = {}

for code in representation_questions:
    representation_results[code] = {}

    for group, columns in subgroup_columns.items():
        result = compare_representation(code, columns)
        representation_results[code][group] = result

        print(f"{code} — {group}")
        display(result)
```

```python
display(representation_results["QC4B1"]["Social grade"])
importance_questions = ["QC4B2", "QC4C2", "QC4D2"]

importance_results = {}

for code in importance_questions:
    importance_results[code] = {}

    for group, columns in subgroup_columns.items():
        result = compare_importance(code, columns)
        importance_results[code][group] = result

        print(f"{code} — {group}")
        display(result)
```

```python

pairs = [
    ("QC4B1", "QC4B2", "Children who look like me"),
    ("QC4C1", "QC4C2", "Children from my region"),
    ("QC4D1", "QC4D2", "Children doing similar activities")
]

rows = []

for availability_code, importance_code, aspect in pairs:
    for group in ["Social grade", "Location"]:
        shortage = representation_results[availability_code][group]["No"]
        importance = importance_results[importance_code][group]["Important"]

        for subgroup, value in shortage.items():
            rows.append({
                "Aspect": aspect,
                "Group": group,
                "Subgroup": subgroup,
                "Measure": "Not enough programmes",
                "Percent": value
            })

        for subgroup, value in importance.items():
            rows.append({
                "Aspect": aspect,
                "Group": group,
                "Subgroup": subgroup,
                "Measure": "Representation is important",
                "Percent": value
            })

viz_data = pd.DataFrame(rows)

#Plot social grade and location separately:

sns.set_theme(style="whitegrid")

group_orders = {
    "Social grade": ["AB", "C1", "C2", "DE"],
    "Location": ["URBAN", "RURAL"]
}

for group, order in group_orders.items():
    g = sns.catplot(
        data=viz_data[viz_data["Group"] == group],
        x="Percent",
        y="Subgroup",
        hue="Measure",
        col="Aspect",
        kind="strip",
        order=order,
        hue_order=[
            "Not enough programmes",
            "Representation is important"
        ],
        palette=["#D97706", "#2563EB"],
        jitter=False,
        dodge=True,
        size=9,
        height=3.5,
        aspect=1,
        legend_out=True
    )

    g.set(
        xlim=(0, 100),
        xticks=range(0, 101, 20)
    )
    g.set_axis_labels("Children (%)", "")
    g.set_titles("{col_name}")
    g.figure.suptitle(
        f"TV representation by {group.lower()}",
        y=1.08
    )

    plt.show()
```

```python

from scipy.stats import linregress

ages = [6, 10, 14]
phone_use = [11, 14, 33]

result = linregress(ages, phone_use)

plt.scatter(ages, phone_use)
plt.plot(
    ages,
    [result.intercept + result.slope * age for age in ages],
    color="red"
)
plt.xlabel("Age-group midpoint")
plt.ylabel("Watching TV or films on a phone (%)")
plt.title("Phone viewing by age group — QP5")
plt.tight_layout()
plt.show()

print(f"Slope: {result.slope:.2f} percentage points per year")
print(f"R²: {result.rvalue**2:.3f}")
```
